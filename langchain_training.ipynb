## 2. Basic LLM Usage {#basic-llm}

Let's start with basic Language Model usage in LangChain.

```python
from langchain.llms import OpenAI
from langchain.chat_models import ChatOpenAI

# Initialize a basic LLM
# Note: You'll need to set your OPENAI_API_KEY environment variable
try:
    llm = OpenAI(temperature=0.7)
    
    # Test the LLM
    response = llm("What is LangChain?")
    print("LLM Response:")
    print(response)
except Exception as e:
    print(f"Error: {e}")
    print("Make sure to set your OPENAI_API_KEY environment variable")
```

## 3. Prompt Templates {#prompt-templates}

Prompt templates allow you to create reusable prompts with variables.

```python
from langchain.prompts import PromptTemplate

# Create a simple prompt template
template = """
You are a helpful assistant that explains concepts clearly.

Question: {question}

Please provide a clear and concise explanation:
"""

prompt = PromptTemplate(
    input_variables=["question"],
    template=template
)

# Format the prompt
formatted_prompt = prompt.format(question="What are the benefits of using LangChain?")
print("Formatted Prompt:")
print(formatted_prompt)

# Use with LLM
try:
    response = llm(formatted_prompt)
    print("\nLLM Response:")
    print(response)
except:
    print("\nLLM not available - showing formatted prompt only")
```

## 4. Chains {#chains}

Chains allow you to combine multiple components together.

```python
from langchain.chains import LLMChain

# Create a chain that combines the prompt template and LLM
try:
    chain = LLMChain(llm=llm, prompt=prompt)
    
    # Run the chain
    result = chain.run(question="How do I get started with machine learning?")
    print("Chain Result:")
    print(result)
except Exception as e:
    print(f"Chain execution failed: {e}")
    print("This requires a valid OpenAI API key")
```

## 5. Memory {#memory}

Memory allows chains to remember previous interactions.

```python
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain

# Create conversation memory
memory = ConversationBufferMemory()

# Create a conversation chain with memory
try:
    conversation = ConversationChain(
        llm=llm,
        memory=memory,
        verbose=True
    )
    
    # Have a conversation
    print("Starting conversation...")
    response1 = conversation.predict(input="Hi, my name is Tony")
    print(f"Response 1: {response1}")
    
    response2 = conversation.predict(input="What's my name?")
    print(f"Response 2: {response2}")
    
except Exception as e:
    print(f"Conversation failed: {e}")
    print("This demonstrates memory structure without actual LLM calls")
    
    # Show memory structure
    print("\nMemory structure:")
    print(f"Memory type: {type(memory)}")
    print(f"Memory variables: {memory.memory_variables}")
```

## 6. Agents {#agents}

Agents can use tools to interact with the external world.

```python
from langchain.agents import Tool, AgentExecutor, create_react_agent
from langchain.tools import DuckDuckGoSearchRun
from langchain import hub

# Create a simple calculator tool
def calculator(expression):
    """Evaluates a mathematical expression"""
    try:
        result = eval(expression)
        return f"The result is: {result}"
    except:
        return "Error: Invalid mathematical expression"

# Create tools list
tools = [
    Tool(
        name="Calculator",
        func=calculator,
        description="Useful for mathematical calculations. Input should be a mathematical expression."
    )
]

print("Available tools:")
for tool in tools:
    print(f"- {tool.name}: {tool.description}")

# Note: Agent creation requires additional setup and API access
print("\nAgent setup complete! (Actual agent execution requires OpenAI API key)")
```

## 7. Document Loading and Processing {#documents}

LangChain provides utilities for loading and processing various document types.

```python
from langchain.text_splitter import CharacterTextSplitter
from langchain.docstore.document import Document

# Create a sample document
sample_text = """
LangChain is a framework for developing applications powered by language models.
It provides a set of tools and abstractions that make it easier to work with LLMs.
The framework includes components for prompt management, chains, memory, agents, and more.
LangChain supports various language models including OpenAI's GPT models, Anthropic's Claude, and others.
It's designed to be modular and extensible, allowing developers to build complex AI applications.
"""

# Create a document
doc = Document(page_content=sample_text, metadata={"source": "langchain_intro"})

# Split the document into chunks
text_splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    separator=" "
)

splits = text_splitter.split_documents([doc])

print(f"Original document length: {len(sample_text)} characters")
print(f"Number of splits: {len(splits)}")
print("\nFirst split:")
print(splits[0].page_content)
print(f"Metadata: {splits[0].metadata}")
```
